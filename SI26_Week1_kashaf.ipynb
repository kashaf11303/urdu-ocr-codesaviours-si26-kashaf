### **1. What is OCR (Optical Character Recognition)?**

Optical Character Recognition (OCR) is a technology that converts text from images, scanned documents, or photographs into editable and searchable digital text. It allows computers to recognize printed or handwritten characters automatically. OCR saves time by removing the need to type text manually from paper documents.

### **2. Why is Urdu OCR harder than English OCR?**

Urdu OCR is more difficult because the Urdu script is cursive, meaning most letters are connected to each other. Many Urdu letters look very similar and only differ by the number or position of dots, making them harder for a computer to distinguish. In addition, Urdu characters change their shape depending on their position in a word, which increases the complexity of accurate recognition.

### **3. What are 2 real-world situations where Urdu OCR would be useful?**

One useful application of Urdu OCR is digitizing old Urdu books, newspapers, and historical documents so they can be searched, edited, and preserved electronically. Another application is automatically extracting text from Urdu forms, invoices, or government documents to reduce manual data entry and improve efficiency in offices and organizations.
